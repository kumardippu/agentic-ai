```python
# @tool decorator
"""This @tool decorator is the simplest way to define a custom tool. The decorator suses the function name as the tool name by defatult, but this 
can be overridden by passing a string as the first argument.
Additionally, the decorator will use th function's docstring as the tool's description - so a docstring must be provided.
"""
from langchain_core.tools import tool

@tool
def division(a: int, b: int) -> int:
    """ Divide two numbers """
    return a / b

print(division.name)
print(division.description)
print(division.args)
```

```python
# async implementation
from langchain_core.tools import tool

@tool
async def amultiply(a: int, b: int) -> int:
    """ Multiply two numbers """
    return a * b
```

```python
from typing import Annotated

@tool
def multiple_by_max(
    a: Annotated[int, "A value"],
    b: Annotated[int, "B value"]) -> int:
    """ Multiply a by the maximum value of b """
    return a* max(b)

print(multiple_by_max.args)
```

```python
# structured tool
"""The StructuredTool.from_function class method provides a bit more configurability than the @tool decorator.
without requiring much additional code.
"""
from types import coroutine
from langchain_core.tools import StructuredTool

def multiply(a: int, b: int) -> int:
    """ Multiply two numbers """
    return a * b

async def amultiply(a: int, b: int) -> int:
    """ Multiply two numbers """
    return a * b

calculator = StructuredTool.from_function(func=multiply, coroutine=amultiply)

print(calculator.invoke({"a": 2, "b": 3}))
print(await calculator.ainvoke({"a": 2, "b": 5}))
```

```python
# Call Tool with LLM model

from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper
from langchain_core.tools import tool
from langchain_tavily import TavilySearch

# 1. Set a custom User-Agent to prevent Wikimedia from blocking the request
wikipedia.set_user_agent("AgenticAIProject/1.0 (contact@example.com)")

# wikipedia tool
api_wrapper = WikipediaAPIWrapper(top_k_results=5, doc_content_chats_max=500)
wiki_tool = WikipediaQueryRun(api_wrapper=api_wrapper)

# tavily search tool
tavily_tool = TavilySearch(
    max_results=5,
    topic="general",
    # include_answer=False,
    # include_raw_content=False,
    # include_images=False,
    # include_image_descriptions=False,
    # search_depth="basic",
    # time_range="day",
    # include_domains=None,
    # exclude_domains=None
)

# custom tools

@tool
def add(a: int, b: int) -> int:
    """Add two numbers together"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers together"""
    return a * b

# list of tools
tools = [wiki_tool, add, multiply, tavily_tool]

tools

# Integrate these tools with an LLM

# initialize model
from langchain.chat_models import init_chat_model

#llm = init_chat_model(model="groq:qwen/qwen3.8-27b", model_provider="groq")
llm = init_chat_model(
    "groq:qwen/qwen3.8-27b"
)


# binding tools to the model
llm_with_tools = llm.bind_tools(tools)
llm_with_tools

from langchain_core.messages import HumanMessage

query = "What is 2* 5"
messages = [HumanMessage(content=query)]

# here content will be blank because LLM is thinking if tools are available and which is the good fit and decide by doc string of the tools
response = llm_with_tools.invoke(query)

response
# check the decided tool call
response.tool_calls

for tool_call in response.tool_calls:
    selected_tool = {"add": add, "multiply": multiply, "wikipedia": wiki_tool, "tavily_search": tavily_tool}[tool_call["name"].lower()]
    tool_msg = selected_tool.invoke(tool_call)
    messages.append(tool_msg)

messages

# with full details
llm_with_tools.invoke(messages)
```
